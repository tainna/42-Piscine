If you want to count letters or numbers, you should put them inside single quotes like 'a', 'b', 'c' or '1', '2', '3'. This is because when you use double quotes like "this", it becomes a string, and the program will not treat it as a single character or number. Also, every character actually represents a number inside the **ASCII table**, which defines what each symbol is. For example, writing a ' ' or the number 32 means the same thing, just like writing 'a' or the number 97 without quotes is the same. To understand it better, you can type man ascii in the terminal to see the full table.

```python
void	ft_print_reverse_alphabet(void)
{
	char	letter;

	letter = 'z';
	while (letter >= 'a')
	{
		write(1, &letter, 1);
		--letter;
	}
}
```

Since we know the program can read letters, we can also count them using functions, like the one above. To make a function that counts, we use the **while** command, which means: *while this condition is true, keep doing what is inside; when it becomes false, stop.* Inside the loop, we can use counters like ++ or -- to increase or decrease values. The difference is that i++ and i-- use the value first and then update it, while ++i and --i update the value first and then use it. In most simple loops like the one above, the result is the same, but it’s good to understand the difference.

```python
#include <unistd.h>

int main(void)
{
    int i = 'a';

    write(1, &i, 1);   // prints 'a'
    write(1, &i++, 1); // prints 'a' (then increases i to 'b')
    write(1, &++i, 1); // increases i to 'c' and then prints 'c'
}
```

**if** and **else** are used to make decisions based on **conditions**. The if part checks if a condition is true, and if it is, the code inside its block runs. The else part runs only when the if condition is false. In your example, the function ft_is_negative checks if n is greater than or equal to 0: if it is, it writes P for positive; otherwise, the else block writes N for negative. This way, the program can choose what to do depending on the value of n.

```python
#include <unistd.h>

void	ft_is_negative(int n)
{
	if (n >= 0)
	{
		write(1, &"P", 1);
	}
	else
	{
		write(1, &"N", 1);
	}
}
```
